```python
from pyspark.sql import functions as F
import datetime
#JSON dizisi (array) olduğu için multiline seçeneğini eklemezsen Spark dosya okurken hata verebilir veya null dönebilir.
employess_raw = spark.read.option("multiline", "true").json('/Volumes/workspace/default/retail_db/employees_raw.json')
display(employess_raw)

bronze_employess= employess_raw.withColumn('ingestion_timestamp', F.current_timestamp())
display(bronze_employess)
```

```python
(
    bronze_employess
    .write
    .format('delta')
    .mode('overwrite')
    .option('overwriteSchema', 'true')
    .saveAsTable('default.bronze_employess')
) 
```
